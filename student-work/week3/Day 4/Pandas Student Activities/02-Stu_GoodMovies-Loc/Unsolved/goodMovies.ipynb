```python
# Dependencies
import pandas as pd
from pathlib import Path 
import re
```

```python
# Load in File from resources
# 'movie_scores.csv'

file = Path('./Resources/movie_scores.csv')
og_df =pd.read_csv(file)
og_df.head()

```

```python
# only care about Imdb, so create a new table that takes the Film and all the columns relating to IMDB
imdb_cols = ["FILM"]
pattern = r".*IMDB.*"

for col in og_df.columns:
    if re.search(pattern, col):
        imdb_cols.append(col)

imbd_df = pd.DataFrame(og_df, columns= imdb_cols)

imbd_df
```

```python
# List only movies whose ratings are over 7 (out of 10) in IMDB

#imbd_df.info()
highly_rated = imbd_df["IMDB"] >= 7.0
imbd_df[highly_rated]
```

```python
imbd_df.info()
```

```python
# Find lesser-known movies to watch, with fewer than 20K votes
less_known = imbd_df["IMDB_user_vote_count"] < 200000
imbd_df[less_known]
```

```python
# Finally, export this file to an Excel spreadsheet -- without the DataFrame index.
imbd_df[less_known].to_excel("output/output.xlsx", sheet_name = "sheet_1", na_rep= " ", header = True, index = False)
```
